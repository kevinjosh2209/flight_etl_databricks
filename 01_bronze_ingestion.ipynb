```python
flight_data = [
    ("AI101", "Mumbai", "Delhi", 120),
    ("AI102", "Delhi", "Bangalore", 140),
    ("AI103", "Mumbai", "Chennai", 110)
]

columns = [
    "flight_no",
    "source",
    "destination",
    "duration"
]

df = spark.createDataFrame(
    flight_data,
    columns
)

display(df)
```

```python
df.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("flights_catalog.bronze.flight_raw")
```
