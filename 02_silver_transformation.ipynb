```python
from pyspark.sql.functions import upper

bronze_table = spark.table("flights_catalog.bronze.flight_raw")

# Example: Convert the 'header' column to uppercase
bronze_df = [col.upper() for col in bronze_table.columns]
silver_df = bronze_table.toDF(*bronze_df)

display(silver_df)
```

```python
silver_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("flights_catalog.silver.flight_clean")
```
